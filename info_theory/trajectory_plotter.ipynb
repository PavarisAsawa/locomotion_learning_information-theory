```python
selected_dataset = terrain
```

```python
output = selected_dataset["flat"][0]["data"]["action_lowpass"][:,0,:]
input = selected_dataset["flat"][0]["data"]["state"][:,0,:]
```

## POS , VEL , NEURAL OUTPUT

```python
num_row = 5
fig, axes = plt.subplots(num_row, 1, figsize=(10, 3*num_row))
# Plot boxplots on each subplot
for i in range(num_row):
    axes[i].plot(input[:, i] , label="pos")
    axes[i].plot(input[:, i+19], label="vel")
    axes[i].plot(input[:, i+19*2], label="action")
    axes[i].set_title(f'Joint {i}')
    axes[i].set_ylabel('Joint value')
    axes[i].legend(loc="upper right", fontsize=8)  # add legend per subplot

plt.tight_layout()
plt.show()
```
